# Customer churn

Practice preprocessing numerical columns and contract categories. [Data notes](../datasets/README.md).


```python
from pathlib import Path

import pandas as pd

root = Path.cwd()
if not (root / "patient_data.csv").exists():
    root = root.parent
data = pd.read_csv(root / "datasets/churn.csv")
print(data.shape)
print(data["label"].value_counts())
data.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

X = data.drop(columns="label")
y = data["label"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
numeric_columns = X.select_dtypes(include="number").columns.tolist()
category_columns = X.select_dtypes(exclude="number").columns.tolist()
preprocess = ColumnTransformer([
    ("numbers", StandardScaler(), numeric_columns),
    ("categories", OneHotEncoder(handle_unknown="ignore"), category_columns),
])
model = make_pipeline(preprocess, LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
prediction = model.predict(X_test)
print("Test accuracy:", accuracy_score(y_test, prediction))
```

```python
pd.crosstab(data["contract"], data["label"])
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, prediction, labels=["churn", "stay"])
plt.title("Churn test predictions")
plt.show()
```
