# Wine classification

Classify the three cultivars in scikit-learn's Wine dataset. The features have different scales.


```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_wine

dataset = load_wine()
X = pd.DataFrame(dataset.data, columns=dataset.feature_names)
y = pd.Series(dataset.target_names[dataset.target], name="label")
print(X.shape)
print(y.value_counts())
X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
prediction = model.predict(X_test)
print("Test accuracy:", accuracy_score(y_test, prediction))
```

```python
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
print("Majority baseline:", accuracy_score(y_test, baseline.predict(X_test)))
```

```python
print(X_train.agg(["mean", "std"]).round(2))
scaled_train = pd.DataFrame(model[0].transform(X_train), columns=X.columns)
print(scaled_train.agg(["mean", "std"]).round(2))
```

```python
row = X_test.iloc[[0]]
print("Actual cultivar:", y_test.loc[row.index].iloc[0])
print("Predicted cultivar:", model.predict(row)[0])
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, prediction, normalize="true", values_format=".2f")
plt.title("Wine test predictions")
plt.show()
```
