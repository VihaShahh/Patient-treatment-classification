# Handwritten digits

Recognize digits from 8 by 8 pixel images bundled with scikit-learn.


```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_digits

dataset = load_digits()
X = pd.DataFrame(dataset.data, columns=dataset.feature_names)
y = pd.Series(dataset.target, name="label")
print(X.shape)
print(y.value_counts())
X.head()
```

```python
plt.imshow(dataset.images[0], cmap="gray")
plt.title(f"Digit: {y.iloc[0]}")
plt.axis("off")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
model = RandomForestClassifier(n_estimators=80, max_depth=12, random_state=42, n_jobs=1)
model.fit(X_train, y_train)
prediction = model.predict(X_test)
print("Test accuracy:", accuracy_score(y_test, prediction))
```

```python
fig, axes = plt.subplots(1, 6, figsize=(10, 3))
for ax, index in zip(axes, X_test.index[:6]):
    ax.imshow(dataset.images[index], cmap="gray")
    label = model.predict(X_test.loc[[index]])[0]
    ax.set_title(f"{y.loc[index]} → {label}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```
