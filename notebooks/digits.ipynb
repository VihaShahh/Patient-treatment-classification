# Handwritten digits

Recognize digits from 8 by 8 pixel images bundled with scikit-learn.


```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_digits

dataset = load_digits()
X = pd.DataFrame(dataset.data, columns=dataset.feature_names)
y = pd.Series(dataset.target, name="label")
print(X.shape)
print(y.value_counts())
X.head()
```

```python
plt.imshow(dataset.images[0], cmap="gray")
plt.title(f"Digit: {y.iloc[0]}")
plt.axis("off")
plt.show()
```
