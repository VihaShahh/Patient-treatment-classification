# Iris flowers

Classify three flower species from four measurements. The data is bundled with scikit-learn.


```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris

dataset = load_iris()
X = pd.DataFrame(dataset.data, columns=dataset.feature_names)
y = pd.Series(dataset.target_names[dataset.target], name="label")
print(X.shape)
print(y.value_counts())
X.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
prediction = model.predict(X_test)
print("Test accuracy:", accuracy_score(y_test, prediction))
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

tree = DecisionTreeClassifier(max_depth=2, random_state=42).fit(X_train, y_train)
plt.figure(figsize=(11, 5))
plot_tree(tree, feature_names=X.columns.tolist(), class_names=tree.classes_.tolist(), filled=True)
plt.show()
```

```python
print("Training accuracy:", model.score(X_train, y_train))
print("Test accuracy:", model.score(X_test, y_test))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
X.iloc[:, :2].plot.box(ax=axes[0], title="Sepal measurements")
X.iloc[:, 2:].plot.box(ax=axes[1], title="Petal measurements")
plt.tight_layout()
plt.show()
```

```python
probabilities = pd.DataFrame(
    model.predict_proba(X_test.head()), columns=model.classes_, index=X_test.head().index
)
probabilities
```

```python
# Measurement order follows the original four feature columns; units are cm.
new_flowers = pd.DataFrame([[5.1, 3.5, 1.4, 0.2], [6.5, 3.0, 5.2, 2.0]], columns=X.columns)
print(model.predict(new_flowers))
```

```python
for label in sorted(y.unique()):
    rows = X[y == label]
    plt.scatter(rows.iloc[:, 2], rows.iloc[:, 3], label=label, s=18)
plt.xlabel(X.columns[2])
plt.ylabel(X.columns[3])
plt.legend()
plt.show()
```
