# Spam classification

Use TF-IDF and logistic regression to classify messages as spam or ham. [Data notes](../datasets/README.md).


```python
from pathlib import Path

import pandas as pd

root = Path.cwd()
if not (root / "patient_data.csv").exists():
    root = root.parent
data = pd.read_csv(root / "datasets/spam.csv")
print(data.shape)
print(data["label"].value_counts())
data.head()
```

```python
from sklearn.model_selection import train_test_split

X = data["message"]
y = data["label"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.pipeline import make_pipeline

model = make_pipeline(TfidfVectorizer(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
prediction = model.predict(X_test)
print(classification_report(y_test, prediction, zero_division=0))
```

```python
examples = pd.DataFrame({"message": X_test, "actual": y_test, "predicted": prediction})
examples[examples["actual"] != examples["predicted"]].head()
```
